### Q26. Does blood pressure relate to age?

```python
df['Sex'] = df['Sex'].map({'Male': 1, 'Female': 0})
df['Blood Pressure'] = pd.to_numeric(df['Blood Pressure'], errors='coerce')
df['Diet'] = df['Diet'].astype('category').cat.codes
df['Country'] = df['Country'].astype('category').cat.codes
df['Continent'] = df['Continent'].astype('category').cat.codes
df['Hemisphere'] = df['Hemisphere'].astype('category').cat.codes


df_corr = df.drop(columns=['Patient ID'])


correlations = df_corr.corr()['Heart Attack Risk'].drop('Heart Attack Risk').sort_values(key=abs, ascending=False)


plt.figure(figsize=(12, 8))
sns.barplot(x=correlations.values, y=correlations.index, palette='coolwarm')
plt.title("Factors Correlated with Heart Attack Risk", fontsize=16)
plt.xlabel("Correlation Coefficient")
plt.ylabel("Features")
plt.grid(axis='x')
plt.tight_layout()
plt.show()
```

```python
sns.histplot(data=df, x='Cholesterol', kde=True)
```

```python
df = df[pd.to_numeric(df['Cholesterol'], errors='coerce').notnull()]
df['Cholesterol'] = df['Cholesterol'].astype(float)

# Plotting boxplot
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='Heart Attack Risk', y='Cholesterol', palette='Set2')
plt.title("Cholesterol Levels by Heart Attack Risk")
plt.xlabel("Heart Attack Risk (0 = No, 1 = Yes)")
plt.ylabel("Cholesterol (mg/dL)")
plt.grid(True)
plt.tight_layout()
plt.show()



```

```python
chol_bins = pd.cut(df['Cholesterol'], bins=[0, 150, 200, 250, 300, 400, 500], labels=[
    'Very Low (<150)', 'Low (150-200)', 'Borderline (200-250)',
    'High (250-300)', 'Very High (300-400)', 'Extreme (>400)'
])


df['Cholesterol Level'] = chol_bins


cholesterol_group = df.groupby('Cholesterol Level')['Heart Attack Risk'].mean().reset_index()


plt.figure(figsize=(10, 6))
sns.barplot(data=cholesterol_group, x='Cholesterol Level', y='Heart Attack Risk', palette='rocket')
plt.title('Heart Attack Risk Across Cholesterol Levels')
plt.xlabel('Cholesterol Level')
plt.ylabel('Average Heart Attack Risk')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['Smoking'] = pd.to_numeric(df['Smoking'], errors='coerce')
df['Alcohol Consumption'] = pd.to_numeric(df['Alcohol Consumption'], errors='coerce')
df['Heart Attack Risk'] = pd.to_numeric(df['Heart Attack Risk'], errors='coerce')


df['Smoker_Alcoholic'] = df.apply(lambda row:
    'Smoker & Alcoholic' if row['Smoking'] == 1 and row['Alcohol Consumption'] == 1 else
    'Smoker Only' if row['Smoking'] == 1 else
    'Alcoholic Only' if row['Alcohol Consumption'] == 1 else
    'Non-Smoker & Non-Alcoholic', axis=1)


plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Smoker_Alcoholic', hue='Heart Attack Risk', palette='Set2')
plt.title("Heart Attack Risk by Smoking & Alcohol Consumption")
plt.xlabel("Category")
plt.ylabel("Number of Patients")
plt.legend(title="Heart Attack Risk", labels=["No", "Yes"])
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
df.rename(columns={'Alcohol Consumption': 'Alcohol Intake'}, inplace=True)


df['Smoking'] = df['Smoking'].astype(int)
df['Alcohol Intake'] = df['Alcohol Intake'].astype(int)
df['Heart Attack Risk'] = df['Heart Attack Risk'].astype(int)


grouped_data = df.groupby(['Smoking', 'Alcohol Intake'])['Heart Attack Risk'].mean().reset_index()


plt.figure(figsize=(10, 6))
sns.barplot(
    data=grouped_data,
    x='Smoking',
    y='Heart Attack Risk',
    hue='Alcohol Intake',
    palette='Set2'
)
plt.title('Effect of Smoking and Alcohol Intake on Heart Attack Risk')
plt.xlabel('Smoking (0 = No, 1 = Yes)')
plt.ylabel('Average Heart Attack Risk')
plt.legend(title='Alcohol Intake (0 = No, 1 = Yes)')
plt.tight_layout()
plt.show()
```

```python
df['Diabetes'] = df['Diabetes'].astype(int)
df['Heart Attack Risk'] = df['Heart Attack Risk'].astype(int)


diabetes_group = df.groupby('Diabetes')['Heart Attack Risk'].mean().reset_index()


plt.figure(figsize=(8, 5))
sns.barplot(data=diabetes_group, x='Diabetes', y='Heart Attack Risk', palette='coolwarm')
plt.title('Effect of Diabetes on Heart Attack Risk')
plt.xlabel('Diabetes (0 = No, 1 = Yes)')
plt.ylabel('Average Heart Attack Risk')
plt.xticks([0, 1], ['Non-Diabetic', 'Diabetic'])
plt.tight_layout()
plt.show()
```

```python
df['Exercises'] = df['Exercise Hours Per Week'].apply(lambda x: 1 if x > 0 else 0)


exercise_binary_group = df.groupby('Exercises')['Heart Attack Risk'].mean().reset_index()


plt.figure(figsize=(8, 5))
sns.barplot(data=exercise_binary_group, x='Exercises', y='Heart Attack Risk', palette='viridis')
plt.title('Heart Attack Risk: Exercisers vs Non-Exercisers')
plt.xlabel('Exercises (0 = No, 1 = Yes)')
plt.ylabel('Average Heart Attack Risk')
plt.xticks([0, 1], ['Non-Exercisers', 'Exercisers'])
plt.tight_layout()
plt.show()
```

```python
bins = [18, 30, 40, 50, 60, 70, 100]
labels = ['18–30', '31–40', '41–50', '51–60', '61–70', '71+']
df['Age Group'] = pd.cut(df['Age'], bins=bins, labels=labels, right=False)


grouped = df.groupby('Age Group')['Heart Attack Risk'].mean().reset_index()


plt.figure(figsize=(8,5))
sns.barplot(x='Age Group', y='Heart Attack Risk', data=grouped, palette='coolwarm')
plt.title('Average Heart Attack Risk by Age Group')
plt.ylabel('Avg Heart Attack Risk')
plt.xlabel('Age Group')
plt.ylim(0, 1)
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
numeric_df = df.select_dtypes(include=[np.number])


features = numeric_df.drop(columns=['Heart Attack Risk'])


correlations = features.corrwith(numeric_df['Heart Attack Risk']).sort_values(ascending=False)


correlation_df = correlations.reset_index()
correlation_df.columns = ['Feature', 'Correlation with Heart Attack Risk']


plt.figure(figsize=(12, 6))
sns.barplot(data=correlation_df, x='Feature', y='Correlation with Heart Attack Risk', palette='coolwarm')
plt.title('Correlation of Features with Heart Attack Risk')
plt.xlabel('Feature')
plt.ylabel('Correlation Coefficient')
plt.xticks(rotation=45)
plt.axhline(0, color='black', linewidth=0.8)  # baseline at 0
plt.tight_layout()
plt.show()
```
